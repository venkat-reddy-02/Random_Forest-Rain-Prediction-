### Preprocessing Of Data

```python
numerical = df.describe().columns      # taking out the numerical columns separately
print(f"Numerical Columns : {numerical}\n")

category = df.describe(include=str).columns     # taking out the category columns
print(f"Category : {category}")
```

```python
# filling Null(Missing) values for numerical features with median

for col in numerical:
    med=df[col].median()
    df[col]=df[col].fillna(med)
```

```python
# filling Null(Missing) values for category features with mode

df['RainToday'] = df['RainToday'].fillna('No')
df['WindDir9am'] = df['WindDir9am'].fillna('N')
df['WindDir3pm'] = df['WindDir3pm'].fillna('SE')
df['WindGustDir'] = df['WindGustDir'].fillna('W')
```

```python
# date feature engineering

df['Date'] = pd.to_datetime(df['Date'])

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
df.drop(['Date'], inplace = True, axis =1)
```

```python
df.drop(['RISK_MM'], inplace = True, axis =1)
```

```python
df.head()
```

```python
# Spliting the data into output(y) and input(X)

X = df.drop('RainTomorrow', axis = 1)
y = df['RainTomorrow']
```

```python
X_train, X_test, y_train, y_test = train_test_split( X, y , test_size = 0.3 , random_state = 21)
```

```python
numerical = X_train.describe().columns
```

```python
# One Hot Encoding on category columns

X_train = pd.concat([X_train[numerical],
                    pd.get_dummies(X_train['Location'], prefix ='Location', dtype = int),
                    pd.get_dummies(X_train['WindGustDir'], prefix = 'WindGustDir', dtype= int),
                    pd.get_dummies(X_train['WindDir9am'], prefix = 'WindDir9am', dtype =int),
                    pd.get_dummies(X_train['WindDir3pm'], prefix = 'WindDir3pm', dtype = int),
                    pd.get_dummies(X_train['RainToday'], prefix = 'RainToday', dtype = int)], axis =1)
```

```python
X_test = pd.concat([X_test[numerical],
                    pd.get_dummies(X_test['Location'], prefix ='Location', dtype = int),
                    pd.get_dummies(X_test['WindGustDir'], prefix = 'WindGustDir', dtype= int),
                    pd.get_dummies(X_test['WindDir9am'], prefix = 'WindDir9am', dtype =int),
                    pd.get_dummies(X_test['WindDir3pm'], prefix = 'WindDir3pm', dtype = int),
                    pd.get_dummies(X_test['RainToday'], prefix = 'RainToday', dtype = int)], axis =1)
```

### Creating Random Forest Model

```python
model = RandomForestClassifier(n_estimators = 150, max_depth= 15, random_state = 28)
```

```python
model.fit(X_train,y_train)
```

```python
y_pred= model.predict(X_test)
```

```python
# Evaluation

accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```
