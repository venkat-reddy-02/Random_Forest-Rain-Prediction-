```python
#filling Null values in X_train,X_test with the mean values of X_train to prevent data leakage

X_train['MinTemp'] = X_train['MinTemp'].fillna(12.185210)
X_train['MaxTemp'] = X_train['MaxTemp'].fillna(23.219571)
X_train['Rainfall'] = X_train['Rainfall'].fillna(0.523628)
X_train['Evaporation'] = X_train['Evaporation'].fillna(1.660964)
X_train['Sunshine'] = X_train['Sunshine'].fillna(7.618194)
X_train['WindGustSpeed'] = X_train['WindGustSpeed'].fillna(3.377062)
X_train['WindSpeed9am'] = X_train['WindSpeed9am'].fillna(14.021725)
X_train['WindSpeed3pm'] = X_train['WindSpeed3pm'].fillna(18.648177)
X_train['Humidity9am'] = X_train['Humidity9am'].fillna(68.867447)
X_train['Humidity3pm'] = X_train['Humidity3pm'].fillna(51.448466)
X_train['Pressure9am'] = X_train['Pressure9am'].fillna(1017.661430)
X_train['Pressure3pm'] = X_train['Pressure3pm'].fillna(1015.270299)
X_train['Cloud9am'] = X_train['Cloud9am'].fillna(4.447843)
X_train['Cloud3pm'] = X_train['Cloud3pm'].fillna(4.508403)
X_train['Temp9am'] = X_train['Temp9am'].fillna(16.979477)
X_train['Temp3pm'] = X_train['Temp3pm'].fillna(21.688582)
X_train['Location'] = X_train['Location'].fillna('Canberra')
X_train['WindGustDir'] = X_train['WindGustDir'].fillna('W')
X_train['WindDir9am'] = X_train['WindDir9am'].fillna('N')
X_train['WindDir3pm'] = X_train['WindDir3pm'].fillna('SE')

X_test['MinTemp'] = X_test['MinTemp'].fillna(12.185210)
X_test['MaxTemp'] = X_test['MaxTemp'].fillna(23.219571)
X_test['Rainfall'] = X_test['Rainfall'].fillna(0.523628)
X_test['Evaporation'] = X_test['Evaporation'].fillna(1.660964)
X_test['Sunshine'] = X_test['Sunshine'].fillna(7.618194)
X_test['WindGustSpeed'] = X_test['WindGustSpeed'].fillna(3.377062)
X_test['WindSpeed9am'] = X_test['WindSpeed9am'].fillna(14.021725)
X_test['WindSpeed3pm'] = X_test['WindSpeed3pm'].fillna(18.648177)
X_test['Humidity9am'] = X_test['Humidity9am'].fillna(68.867447)
X_test['Humidity3pm'] = X_test['Humidity3pm'].fillna(51.448466)
X_test['Pressure9am'] = X_test['Pressure9am'].fillna(1017.661430)
X_test['Pressure3pm'] = X_test['Pressure3pm'].fillna(1015.270299)
X_test['Cloud9am'] = X_test['Cloud9am'].fillna(4.447843)
X_test['Cloud3pm'] = X_test['Cloud3pm'].fillna(4.508403)
X_test['Temp9am'] = X_test['Temp9am'].fillna(16.979477)
X_test['Temp3pm'] = X_test['Temp3pm'].fillna(21.688582)
X_test['Location'] = X_test['Location'].fillna('Canberra')
X_test['WindGustDir'] = X_test['WindGustDir'].fillna('W')
X_test['WindDir9am'] = X_test['WindDir9am'].fillna('N')
X_test['WindDir3pm'] = X_test['WindDir3pm'].fillna('SE')
```

```python

X_train = pd.concat([ X_train[numerical],
                      pd.get_dummies(X_train['Location'],prefix='Loc',drop_first=True),
                      pd.get_dummies(X_train['WindGustDir'],prefix='WGD',drop_first=True),
                      pd.get_dummies(X_train['WindDir9am'],prefix='WD9',drop_first=True),
                      pd.get_dummies(X_train['WindDir3pm'],prefix='WD3',drop_first=True),
                      pd.get_dummies(X_train['RainToday'],drop_first=True)],axis = 1)
```

```python
X_test = pd.concat([ X_test[numerical],
                      pd.get_dummies(X_test['Location'],prefix='Loc',drop_first=True),
                      pd.get_dummies(X_test['WindGustDir'],prefix='WGD',drop_first=True),
                      pd.get_dummies(X_test['WindDir9am'],prefix='WD9',drop_first=True),
                      pd.get_dummies(X_test['WindDir3pm'],prefix='WD3',drop_first=True),
                      pd.get_dummies(X_test['RainToday'],drop_first=True)],axis = 1)
```

```python
for  depth in [11,12,13,14,16,17]:
    model1 = RandomForestClassifier(
        n_estimators= 32,
        max_depth = depth,
        random_state = 23
    )
    model1.fit(X_train,y_train)
    y_pred = model1.predict(X_test)
    y_train_pred = model1.predict(X_train)
    print(f"Test Accuracy = {accuracy_score(y_test,y_pred)}, Train accuracy = {accuracy_score(y_train,y_train_pred)} At max_depth of {depth}")
```
